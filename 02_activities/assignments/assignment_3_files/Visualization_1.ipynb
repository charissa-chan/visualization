# Visualization 1

Dataset used: Deaths of Shelter Residents
https://open.toronto.ca/dataset/deaths-of-shelter-residents/

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
df = pd.read_csv("/Users/charissachan/Desktop/DSI/visualization/02_activities/assignments/assigment_3_files/Deaths of Shelter Residents.csv")
df
```

```python
df_2010_2025 = df[(df["Year"] >= 2010) & (df["Year"] <= 2025)]

df_2010_2025
```

```python
Yearly = df_2010_2025.groupby("Year", as_index=False)[["Male", "Female", "Total decedents"]].sum()
```

```python
plt.style.use('fast')
fig, ax = plt.subplots(figsize=(15, 6))

x = Yearly["Year"]
y_male = Yearly["Male"]
y_female = Yearly["Female"]
y_total = Yearly["Total decedents"]
width = 0.4

ax.bar(x - 0.2, y_male, width, label="Male Deaths")
ax.bar(x + 0.2, y_female, width, label="Female Deaths")

ax.plot(x, y_total, marker="o", linewidth=2, label="Total Deaths")

ax.set_title("Deaths of Toronto Shelter Residents (2010–2025)", fontsize=25)
ax.set_xlabel("Year")
ax.set_ylabel("Number of Deaths")


ax.set_xticks(x)
ax.set_xticklabels(x.astype(str), rotation=45)


ax.legend(loc = 'upper left', 
       bbox_to_anchor = (1, 1))

ax.margins(x=0.02)
ax.margins(y=0.1)

plt.tight_layout()
plt.savefig("visualization_1.png")
plt.show()
```
