### Data Visualization: Subplots and Combining Visualizations

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

### Activity: Customizing our plots

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])

font1 = {'family': 'serif', 'color': 'teal'}
font2 = {'family': 'serif', 'color': 'sienna'}

fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(7, 3))
ax1.scatter(x1, y1, color='cornflowerblue')
ax1.set_xlabel('Random X Axis Label!', fontsize=12, fontdict=font1)
ax1.set_title('Scatter Plot of Random Values', fontsize=10, color='black')

ax2.bar(x2, y2, color='sandybrown') # light salmon (soft orange-peach)
ax2.set_xlabel('Random X Axis Label!', fontsize=12, fontdict=font2)
ax2.set_title('Bar Chart of Random Values', fontsize=10, color='black')
plt.show()
```
